```python
# ===============================================
# 0. Imports
# ===============================================
import os
import zipfile
from glob import glob
import cv2
import numpy as np
import matplotlib.pyplot as plt
from tqdm import tqdm
import tensorflow as tf
from tensorflow.keras import layers, models
from sklearn.model_selection import train_test_split
import tensorflow.keras.backend as K

# ===============================================
# 1. Unzip Dataset + Auto Path Fix
# ===============================================
zip_path = "/content/stage1_train.zip"
extract_root = "/content/stage1_train"

# ✅ Make sure extraction root exists
os.makedirs(extract_root, exist_ok=True)

# Check if dataset already extracted
if not any(os.path.isdir(os.path.join(extract_root, d)) for d in os.listdir(extract_root)):
    print("📦 Extracting dataset...")
    with zipfile.ZipFile(zip_path, 'r') as zip_ref:
        zip_ref.extractall(extract_root)
    print("✅ Extraction completed.")
else:
    print("✅ Dataset already extracted.")

# Find candidate folders
candidates = [d for d in glob(os.path.join(extract_root, "*")) if os.path.isdir(d)]
if len(candidates) == 0:
    raise FileNotFoundError("❌ Could not find extracted stage1_train folder after extraction.")

# Handle nested stage1_train/stage1_train issue
if len(candidates) == 1 and "stage1_train" in os.path.basename(candidates[0]):
    extract_path = candidates[0]
    sub_candidates = [d for d in glob(os.path.join(extract_path, "*")) if os.path.isdir(d)]
    if len(sub_candidates) > 0:
        extract_path = extract_path  # keep inside nested folder
else:
    extract_path = extract_root

print("✅ Using dataset folder:", extract_path)
print("🔍 First 5 contents:", os.listdir(extract_path)[:5])

# ===============================================
# 2. Load Data Function (Recursive)
# ===============================================
def load_data(data_dir, img_size=(224,224)):
    images, masks = [], []

    # Recursively find patient folders
    all_ids = [d for d in glob(os.path.join(data_dir, "*")) if os.path.isdir(d)]
    print(f"🔍 Found {len(all_ids)} patient folders")

    for img_id in tqdm(all_ids, desc="Loading patients"):
        img_path_list = glob(os.path.join(img_id, "images", "*.png"))
        mask_paths = glob(os.path.join(img_id, "masks", "*.png"))

        if len(img_path_list) == 0 or len(mask_paths) == 0:
            continue  # skip empty or invalid folders

        # Load image
        image = cv2.imread(img_path_list[0], cv2.IMREAD_COLOR)
        if image is None:
            continue
        image = cv2.resize(image, img_size)
        image = image / 255.0

        # Merge all masks
        mask = np.zeros(img_size, dtype=np.uint8)
        for mp in mask_paths:
            m = cv2.imread(mp, cv2.IMREAD_GRAYSCALE)
            if m is None:
                continue
            m = cv2.resize(m, img_size)
            mask = np.maximum(mask, m)
        mask = (mask > 127).astype(np.float32)
        mask = np.expand_dims(mask, axis=-1)

        images.append(image)
        masks.append(mask)

    return np.array(images, dtype=np.float32), np.array(masks, dtype=np.float32)

# ===============================================
# 3. Load Dataset
# ===============================================
IMG_SIZE = (224,224)
images, masks = load_data(extract_path, img_size=IMG_SIZE)
print("✅ Images:", images.shape)
print("✅ Masks:", masks.shape)

if len(images) == 0:
    raise RuntimeError("❌ No images loaded! Check dataset structure inside the zip.")

# Quick visualization
idx = np.random.randint(0, len(images))
plt.subplot(1,2,1); plt.imshow(images[idx]); plt.title("Image"); plt.axis("off")
plt.subplot(1,2,2); plt.imshow(masks[idx].squeeze(), cmap="gray"); plt.title("Mask"); plt.axis("off")
plt.show()

# ===============================================
# 4. Loss Functions
# ===============================================
def dice_loss(y_true, y_pred, smooth=1e-6):
    y_true_f = K.flatten(y_true)
    y_pred_f = K.flatten(y_pred)
    intersection = K.sum(y_true_f * y_pred_f)
    return 1 - (2. * intersection + smooth) / (K.sum(y_true_f) + K.sum(y_pred_f) + smooth)

def bce_dice_loss(y_true, y_pred):
    bce = tf.keras.losses.binary_crossentropy(y_true, y_pred)
    dice = dice_loss(y_true, y_pred)
    return bce + dice

# ===============================================
# 5. TransUNet Model
# ===============================================
class PatchEmbedding(layers.Layer):
    def __init__(self, embed_dim=768):
        super().__init__()
        self.proj = layers.Conv2D(embed_dim, kernel_size=1, strides=1, padding="valid")
        self.flatten = layers.Reshape((-1, embed_dim))

    def call(self, x):
        x = self.proj(x)
        x = self.flatten(x)
        return x

class TransformerBlock(layers.Layer):
    def __init__(self, embed_dim=768, num_heads=12, mlp_dim=3072, dropout=0.1):
        super().__init__()
        self.norm1 = layers.LayerNormalization(epsilon=1e-6)
        self.attn = layers.MultiHeadAttention(num_heads=num_heads, key_dim=embed_dim, dropout=dropout)
        self.drop1 = layers.Dropout(dropout)
        self.norm2 = layers.LayerNormalization(epsilon=1e-6)
        self.mlp = models.Sequential([
            layers.Dense(mlp_dim, activation=tf.nn.gelu),
            layers.Dropout(dropout),
            layers.Dense(embed_dim),
            layers.Dropout(dropout),
        ])

    def call(self, x):
        attn_output = self.attn(x, x)
        x = x + self.drop1(attn_output)
        mlp_output = self.mlp(self.norm2(x))
        return x + mlp_output

def build_transformer_encoder(x, num_layers=4, embed_dim=768, num_heads=12, mlp_dim=3072):
    # Fewer layers for faster training
    for _ in range(num_layers):
        x = TransformerBlock(embed_dim, num_heads, mlp_dim)(x)
    return x

def build_transunet(input_shape=(224,224,3), num_classes=1):
    inputs = layers.Input(shape=input_shape)
    base_model = tf.keras.applications.ResNet50(include_top=False, weights="imagenet", input_tensor=inputs)

    skip1 = base_model.get_layer("conv1_relu").output
    skip2 = base_model.get_layer("conv2_block3_out").output
    skip3 = base_model.get_layer("conv3_block4_out").output
    cnn_feature = base_model.get_layer("conv4_block6_out").output

    patches = PatchEmbedding(embed_dim=768)(cnn_feature)
    transformer_out = build_transformer_encoder(patches)
    h, w = cnn_feature.shape[1], cnn_feature.shape[2]
    x = layers.Reshape((h, w, 768))(transformer_out)

    x = layers.Conv2DTranspose(512, 3, strides=2, padding="same")(x)
    x = layers.Concatenate()([x, skip3])
    x = layers.Conv2D(512, 3, padding="same", activation="relu")(x)

    x = layers.Conv2DTranspose(256, 3, strides=2, padding="same")(x)
    x = layers.Concatenate()([x, skip2])
    x = layers.Conv2D(256, 3, padding="same", activation="relu")(x)

    x = layers.Conv2DTranspose(128, 3, strides=2, padding="same")(x)
    x = layers.Concatenate()([x, skip1])
    x = layers.Conv2D(128, 3, padding="same", activation="relu")(x)

    x = layers.Conv2DTranspose(64, 3, strides=2, padding="same")(x)
    x = layers.Conv2D(64, 3, padding="same", activation="relu")(x)

    outputs = layers.Conv2D(num_classes, 1, activation="sigmoid")(x)
    return models.Model(inputs, outputs, name="TransUNet")

# ===============================================
# 6. Train/Validation Split
# ===============================================
X_train, X_val, y_train, y_val = train_test_split(images, masks, test_size=0.2, random_state=42)
print("✅ Final Train:", X_train.shape, y_train.shape)
print("✅ Final Val:", X_val.shape, y_val.shape)

# ===============================================
# 7. Compile Model
# ===============================================
model = build_transunet(input_shape=(224,224,3))
model.compile(optimizer=tf.keras.optimizers.Adam(1e-4),
              loss=bce_dice_loss,
              metrics=["accuracy"])
model.summary()

# ===============================================
# 8. Callbacks
# ===============================================
callbacks = [
    tf.keras.callbacks.EarlyStopping(monitor="val_loss", patience=10, restore_best_weights=True),
    tf.keras.callbacks.ModelCheckpoint("transunet_dsb.h5", save_best_only=True, monitor="val_loss")
]

# ===============================================
# 9. Train
# ===============================================
history = model.fit(
    X_train, y_train,
    validation_data=(X_val, y_val),
    epochs=50,
    batch_size=8,
    callbacks=callbacks
)

# ===============================================
# 10. Evaluate & Visualize
# ===============================================
plt.plot(history.history["loss"], label="train_loss")
plt.plot(history.history["val_loss"], label="val_loss")
plt.legend(); plt.show()

idx = np.random.randint(0, len(X_val))
pred = model.predict(np.expand_dims(X_val[idx], axis=0))[0,:,:,0]

plt.subplot(1,3,1); plt.imshow(X_val[idx]); plt.title("Image"); plt.axis("off")
plt.subplot(1,3,2); plt.imshow(y_val[idx].squeeze(), cmap="gray"); plt.title("Ground Truth"); plt.axis("off")
plt.subplot(1,3,3); plt.imshow(pred > 0.5, cmap="gray"); plt.title("Prediction"); plt.axis("off")
plt.show()
```

```python
# ===============================================
# 11. Evaluate Segmentation Accuracy
# ===============================================
def calculate_metrics(y_true, y_pred, threshold=0.5, smooth=1e-6):
    """
    y_true, y_pred: arrays of shape (N, H, W, 1)
    """
    y_pred_bin = (y_pred > threshold).astype(np.float32)

    # Pixel-wise accuracy
    pixel_acc = np.mean(y_true == y_pred_bin)

    # Dice Score
    intersection = np.sum(y_true * y_pred_bin)
    dice = (2 * intersection + smooth) / (np.sum(y_true) + np.sum(y_pred_bin) + smooth)

    # IoU
    union = np.sum(y_true) + np.sum(y_pred_bin) - intersection
    iou = (intersection + smooth) / (union + smooth)

    return pixel_acc, dice, iou

# Predict on validation set
preds_val = model.predict(X_val)

pixel_acc, dice_score, iou_score = calculate_metrics(y_val, preds_val)

print(f"Pixel-wise Accuracy: {pixel_acc:.4f}")
print(f"Dice Score: {dice_score:.4f}")
print(f"IoU Score: {iou_score:.4f}")
```

```python
# ===============================================
# 0. Imports
# ===============================================
import os
import zipfile
from glob import glob
import cv2
import numpy as np
import matplotlib.pyplot as plt
from tqdm import tqdm
import tensorflow as tf
from tensorflow.keras import layers, models
from sklearn.model_selection import train_test_split
import tensorflow.keras.backend as K

# ✅ Enable mixed precision (helps with large 512x512 models)
from tensorflow.keras import mixed_precision
mixed_precision.set_global_policy("mixed_float16")

# ===============================================
# 1. Unzip Dataset + Auto Path Fix
# ===============================================
zip_path = "/content/stage1_train.zip"
extract_root = "/content/stage1_train"
os.makedirs(extract_root, exist_ok=True)

if not any(os.path.isdir(os.path.join(extract_root, d)) for d in os.listdir(extract_root)):
    print("📦 Extracting dataset...")
    with zipfile.ZipFile(zip_path, 'r') as zip_ref:
        zip_ref.extractall(extract_root)
    print("✅ Extraction completed.")
else:
    print("✅ Dataset already extracted.")

candidates = [d for d in glob(os.path.join(extract_root, "*")) if os.path.isdir(d)]
if len(candidates) == 0:
    raise FileNotFoundError("❌ Could not find extracted stage1_train folder.")

if len(candidates) == 1 and "stage1_train" in os.path.basename(candidates[0]):
    extract_path = candidates[0]
else:
    extract_path = extract_root

print("✅ Using dataset folder:", extract_path)
print("🔍 First 5 contents:", os.listdir(extract_path)[:5])

# ===============================================
# 2. Load Data Function (Recursive)
# ===============================================
def load_data(data_dir, img_size=(512, 512)):
    images, masks = [], []
    all_ids = [d for d in glob(os.path.join(data_dir, "*")) if os.path.isdir(d)]
    print(f"🔍 Found {len(all_ids)} patient folders")

    for img_id in tqdm(all_ids, desc="Loading patients"):
        img_path_list = glob(os.path.join(img_id, "images", "*.png"))
        mask_paths = glob(os.path.join(img_id, "masks", "*.png"))

        if len(img_path_list) == 0 or len(mask_paths) == 0:
            continue

        image = cv2.imread(img_path_list[0], cv2.IMREAD_COLOR)
        if image is None:
            continue
        image = cv2.resize(image, img_size)
        image = image / 255.0

        mask = np.zeros(img_size, dtype=np.uint8)
        for mp in mask_paths:
            m = cv2.imread(mp, cv2.IMREAD_GRAYSCALE)
            if m is None:
                continue
            m = cv2.resize(m, img_size)
            mask = np.maximum(mask, m)
        mask = (mask > 127).astype(np.float32)
        mask = np.expand_dims(mask, axis=-1)

        images.append(image)
        masks.append(mask)

    return np.array(images, dtype=np.float32), np.array(masks, dtype=np.float32)

# ===============================================
# 3. Load Dataset
# ===============================================
IMG_SIZE = (512, 512)
images, masks = load_data(extract_path, img_size=IMG_SIZE)
print("✅ Images:", images.shape)
print("✅ Masks:", masks.shape)

if len(images) == 0:
    raise RuntimeError("❌ No images loaded!")

idx = np.random.randint(0, len(images))
plt.subplot(1,2,1); plt.imshow(images[idx]); plt.title("Image"); plt.axis("off")
plt.subplot(1,2,2); plt.imshow(masks[idx].squeeze(), cmap="gray"); plt.title("Mask"); plt.axis("off")
plt.show()

# ===============================================
# 4. Loss Functions
# ===============================================
def dice_loss(y_true, y_pred, smooth=1e-6):
    y_true_f = K.flatten(y_true)
    y_pred_f = K.flatten(y_pred)
    intersection = K.sum(y_true_f * y_pred_f)
    return 1 - (2. * intersection + smooth) / (K.sum(y_true_f) + K.sum(y_pred_f) + smooth)

def bce_dice_loss(y_true, y_pred):
    bce = tf.keras.losses.binary_crossentropy(y_true, y_pred)
    dice = dice_loss(y_true, y_pred)
    return bce + dice

# ===============================================
# 5. TransUNet Model with Positional Encoding
# ===============================================
class PatchEmbedding(layers.Layer):
    def __init__(self, embed_dim=768):
        super().__init__()
        self.proj = layers.Conv2D(embed_dim, kernel_size=1, strides=1, padding="valid")
        self.flatten = layers.Reshape((-1, embed_dim))

    def call(self, x):
        x = self.proj(x)
        x = self.flatten(x)
        return x

class PositionalEncoding(layers.Layer):
    def __init__(self, num_patches, embed_dim):
        super().__init__()
        self.pos_embed = self.add_weight(
            name="pos_embed",
            shape=(1, num_patches, embed_dim),
            initializer="random_normal"
        )

    def call(self, x):
        return x + self.pos_embed

class TransformerBlock(layers.Layer):
    def __init__(self, embed_dim=768, num_heads=12, mlp_dim=3072, dropout=0.1):
        super().__init__()
        self.norm1 = layers.LayerNormalization(epsilon=1e-6)
        self.attn = layers.MultiHeadAttention(num_heads=num_heads, key_dim=embed_dim, dropout=dropout)
        self.drop1 = layers.Dropout(dropout)
        self.norm2 = layers.LayerNormalization(epsilon=1e-6)
        self.mlp = models.Sequential([
            layers.Dense(mlp_dim, activation=tf.nn.gelu),
            layers.Dropout(dropout),
            layers.Dense(embed_dim),
            layers.Dropout(dropout),
        ])

    def call(self, x):
        attn_output = self.attn(self.norm1(x), self.norm1(x))
        x = x + self.drop1(attn_output)
        mlp_output = self.mlp(self.norm2(x))
        return x + mlp_output

def build_transformer_encoder(x, num_layers=4, embed_dim=768, num_heads=12, mlp_dim=3072):
    for _ in range(num_layers):
        x = TransformerBlock(embed_dim, num_heads, mlp_dim)(x)
    return x

def build_transunet(input_shape=(512, 512, 3), num_classes=1):
    inputs = layers.Input(shape=input_shape)

    base_model = tf.keras.applications.ResNet50(
        include_top=False,
        weights="imagenet",
        input_tensor=inputs
    )

    skip1 = base_model.get_layer("conv1_relu").output          # 256x256
    skip2 = base_model.get_layer("conv2_block3_out").output    # 128x128
    skip3 = base_model.get_layer("conv3_block4_out").output    # 64x64
    cnn_feature = base_model.get_layer("conv4_block6_out").output  # 32x32

    # Tokenize features
    patches = PatchEmbedding(embed_dim=768)(cnn_feature)

    # Add positional encoding
    num_patches = patches.shape[1]
    patches = PositionalEncoding(num_patches, 768)(patches)

    # Transformer encoder
    transformer_out = build_transformer_encoder(patches)

    # Reshape back to feature map
    h, w = cnn_feature.shape[1], cnn_feature.shape[2]
    x = layers.Reshape((h, w, 768))(transformer_out)

    # Decoder
    x = layers.Conv2DTranspose(512, 3, strides=2, padding="same")(x)
    x = layers.Concatenate()([x, skip3])
    x = layers.Conv2D(512, 3, padding="same", activation="relu")(x)

    x = layers.Conv2DTranspose(256, 3, strides=2, padding="same")(x)
    x = layers.Concatenate()([x, skip2])
    x = layers.Conv2D(256, 3, padding="same", activation="relu")(x)

    x = layers.Conv2DTranspose(128, 3, strides=2, padding="same")(x)
    x = layers.Concatenate()([x, skip1])
    x = layers.Conv2D(128, 3, padding="same", activation="relu")(x)

    x = layers.Conv2DTranspose(64, 3, strides=2, padding="same")(x)
    x = layers.Conv2D(64, 3, padding="same", activation="relu")(x)

    outputs = layers.Conv2D(num_classes, 1, activation="sigmoid", dtype="float32")(x)
    return models.Model(inputs, outputs, name="TransUNet")

# ===============================================
# 6. Train/Validation Split
# ===============================================
X_train, X_val, y_train, y_val = train_test_split(images, masks, test_size=0.2, random_state=42)
print("✅ Final Train:", X_train.shape, y_train.shape)
print("✅ Final Val:", X_val.shape, y_val.shape)

# ===============================================
# 7. Compile Model
# ===============================================
model = build_transunet(input_shape=(512,512,3))
model.compile(optimizer=tf.keras.optimizers.Adam(1e-4),
              loss=bce_dice_loss,
              metrics=["accuracy"])
model.summary()

# ===============================================
# 8. Callbacks
# ===============================================
callbacks = [
    tf.keras.callbacks.ModelCheckpoint("transunet_pe.h5", save_best_only=True, monitor="val_loss")
]

# ===============================================
# 9. Train
# ===============================================
history = model.fit(
    X_train, y_train,
    validation_data=(X_val, y_val),
    epochs=50,
    batch_size=8,   # reduce batch size for 512x512
    callbacks=callbacks
)

# ===============================================
# 10. Evaluate & Visualize
# ===============================================
plt.plot(history.history["loss"], label="train_loss")
plt.plot(history.history["val_loss"], label="val_loss")
plt.legend(); plt.show()

idx = np.random.randint(0, len(X_val))
pred = model.predict(np.expand_dims(X_val[idx], axis=0))[0,:,:,0]

plt.subplot(1,3,1); plt.imshow(X_val[idx]); plt.title("Image"); plt.axis("off")
plt.subplot(1,3,2); plt.imshow(y_val[idx].squeeze(), cmap="gray"); plt.title("Ground Truth"); plt.axis("off")
plt.subplot(1,3,3); plt.imshow(pred > 0.5, cmap="gray"); plt.title("Prediction"); plt.axis("off")
plt.show()

# ===============================================
# 11. Evaluate Segmentation Accuracy
# ===============================================
def calculate_metrics(y_true, y_pred, threshold=0.5, smooth=1e-6):
    y_pred_bin = (y_pred > threshold).astype(np.float32)
    pixel_acc = np.mean(y_true == y_pred_bin)
    intersection = np.sum(y_true * y_pred_bin)
    dice = (2 * intersection + smooth) / (np.sum(y_true) + np.sum(y_pred_bin) + smooth)
    union = np.sum(y_true) + np.sum(y_pred_bin) - intersection
    iou = (intersection + smooth) / (union + smooth)
    return pixel_acc, dice, iou

preds_val = model.predict(X_val)
pixel_acc, dice_score, iou_score = calculate_metrics(y_val, preds_val)

print(f"Pixel-wise Accuracy: {pixel_acc:.4f}")
print(f"Dice Score: {dice_score:.4f}")
print(f"IoU Score: {iou_score:.4f}")
```

```python
# ===============================================
# 0. Imports
# ===============================================
import os
import zipfile
from glob import glob
import cv2
import numpy as np
import matplotlib.pyplot as plt
from tqdm import tqdm
import tensorflow as tf
from tensorflow.keras import layers, models
from sklearn.model_selection import KFold
import tensorflow.keras.backend as K

# ✅ Enable mixed precision
from tensorflow.keras import mixed_precision
mixed_precision.set_global_policy("mixed_float16")

# ===============================================
# 1. Unzip Dataset + Auto Path Fix
# ===============================================
zip_path = "/content/stage1_train.zip"
extract_root = "/content/stage1_train"
os.makedirs(extract_root, exist_ok=True)

if not any(os.path.isdir(os.path.join(extract_root, d)) for d in os.listdir(extract_root)):
    print("📦 Extracting dataset...")
    with zipfile.ZipFile(zip_path, 'r') as zip_ref:
        zip_ref.extractall(extract_root)
    print("✅ Extraction completed.")
else:
    print("✅ Dataset already extracted.")

candidates = [d for d in glob(os.path.join(extract_root, "*")) if os.path.isdir(d)]
if len(candidates) == 0:
    raise FileNotFoundError("❌ Could not find extracted stage1_train folder.")

if len(candidates) == 1 and "stage1_train" in os.path.basename(candidates[0]):
    extract_path = candidates[0]
else:
    extract_path = extract_root

print("✅ Using dataset folder:", extract_path)
print("🔍 First 5 contents:", os.listdir(extract_path)[:5])

# ===============================================
# 2. Load Data Function (Recursive)
# ===============================================
def load_data(data_dir, img_size=(512, 512)):
    images, masks = [], []
    all_ids = [d for d in glob(os.path.join(data_dir, "*")) if os.path.isdir(d)]
    print(f"🔍 Found {len(all_ids)} patient folders")

    for img_id in tqdm(all_ids, desc="Loading patients"):
        img_path_list = glob(os.path.join(img_id, "images", "*.png"))
        mask_paths = glob(os.path.join(img_id, "masks", "*.png"))

        if len(img_path_list) == 0 or len(mask_paths) == 0:
            continue

        image = cv2.imread(img_path_list[0], cv2.IMREAD_COLOR)
        if image is None:
            continue
        image = cv2.resize(image, img_size)
        image = image / 255.0

        mask = np.zeros(img_size, dtype=np.uint8)
        for mp in mask_paths:
            m = cv2.imread(mp, cv2.IMREAD_GRAYSCALE)
            if m is None:
                continue
            m = cv2.resize(m, img_size)
            mask = np.maximum(mask, m)
        mask = (mask > 127).astype(np.float32)
        mask = np.expand_dims(mask, axis=-1)

        images.append(image)
        masks.append(mask)

    return np.array(images, dtype=np.float32), np.array(masks, dtype=np.float32)

# ===============================================
# 3. Load Dataset
# ===============================================
IMG_SIZE = (512, 512)
images, masks = load_data(extract_path, img_size=IMG_SIZE)
print("✅ Images:", images.shape)
print("✅ Masks:", masks.shape)

if len(images) == 0:
    raise RuntimeError("❌ No images loaded!")

idx = np.random.randint(0, len(images))
plt.subplot(1, 2, 1); plt.imshow(images[idx]); plt.title("Image"); plt.axis("off")
plt.subplot(1, 2, 2); plt.imshow(masks[idx].squeeze(), cmap="gray"); plt.title("Mask"); plt.axis("off")
plt.show()

# ===============================================
# 4. Loss Functions
# ===============================================
def dice_loss(y_true, y_pred, smooth=1e-6):
    y_true_f = K.flatten(y_true)
    y_pred_f = K.flatten(y_pred)
    intersection = K.sum(y_true_f * y_pred_f)
    return 1 - (2. * intersection + smooth) / (K.sum(y_true_f) + K.sum(y_pred_f) + smooth)

def bce_dice_loss(y_true, y_pred):
    bce = tf.keras.losses.binary_crossentropy(y_true, y_pred)
    dice = dice_loss(y_true, y_pred)
    return bce + dice

# ===============================================
# 5. TransUNet Model
# ===============================================
class PatchEmbedding(layers.Layer):
    def __init__(self, embed_dim=768):
        super().__init__()
        self.proj = layers.Conv2D(embed_dim, kernel_size=1, strides=1, padding="valid")
        self.flatten = layers.Reshape((-1, embed_dim))

    def call(self, x):
        x = self.proj(x)
        x = self.flatten(x)
        return x

class PositionalEncoding(layers.Layer):
    def __init__(self, num_patches, embed_dim):
        super().__init__()
        self.pos_embed = self.add_weight(
            name="pos_embed",
            shape=(1, num_patches, embed_dim),
            initializer="random_normal"
        )

    def call(self, x):
        return x + self.pos_embed

class TransformerBlock(layers.Layer):
    def __init__(self, embed_dim=768, num_heads=12, mlp_dim=3072, dropout=0.1):
        super().__init__()
        self.norm1 = layers.LayerNormalization(epsilon=1e-6)
        self.attn = layers.MultiHeadAttention(num_heads=num_heads, key_dim=embed_dim, dropout=dropout)
        self.drop1 = layers.Dropout(dropout)
        self.norm2 = layers.LayerNormalization(epsilon=1e-6)
        self.mlp = models.Sequential([
            layers.Dense(mlp_dim, activation=tf.nn.gelu),
            layers.Dropout(dropout),
            layers.Dense(embed_dim),
            layers.Dropout(dropout),
        ])

    def call(self, x):
        attn_output = self.attn(self.norm1(x), self.norm1(x))
        x = x + self.drop1(attn_output)
        mlp_output = self.mlp(self.norm2(x))
        return x + mlp_output

def build_transformer_encoder(x, num_layers=4, embed_dim=768, num_heads=12, mlp_dim=3072):
    for _ in range(num_layers):
        x = TransformerBlock(embed_dim, num_heads, mlp_dim)(x)
    return x

def build_transunet(input_shape=(512, 512, 3), num_classes=1):
    inputs = layers.Input(shape=input_shape)

    base_model = tf.keras.applications.ResNet50(
        include_top=False,
        weights="imagenet",
        input_tensor=inputs
    )

    skip1 = base_model.get_layer("conv1_relu").output          # 256x256
    skip2 = base_model.get_layer("conv2_block3_out").output    # 128x128
    skip3 = base_model.get_layer("conv3_block4_out").output    # 64x64
    cnn_feature = base_model.get_layer("conv4_block6_out").output  # 32x32

    # Tokenize features
    patches = PatchEmbedding(embed_dim=768)(cnn_feature)

    # Add positional encoding
    num_patches = patches.shape[1]
    patches = PositionalEncoding(num_patches, 768)(patches)

    # Transformer encoder
    transformer_out = build_transformer_encoder(patches)

    # Reshape back to feature map
    h, w = cnn_feature.shape[1], cnn_feature.shape[2]
    x = layers.Reshape((h, w, 768))(transformer_out)

    # Decoder
    x = layers.Conv2DTranspose(512, 3, strides=2, padding="same")(x)
    x = layers.Concatenate()([x, skip3])
    x = layers.Conv2D(512, 3, padding="same", activation=tf.nn.leaky_relu)(x)

    x = layers.Conv2DTranspose(256, 3, strides=2, padding="same")(x)
    x = layers.Concatenate()([x, skip2])
    x = layers.Conv2D(256, 3, padding="same", activation=tf.nn.leaky_relu)(x)

    x = layers.Conv2DTranspose(128, 3, strides=2, padding="same")(x)
    x = layers.Concatenate()([x, skip1])
    x = layers.Conv2D(128, 3, padding="same", activation=tf.nn.leaky_relu)(x)

    x = layers.Conv2DTranspose(64, 3, strides=2, padding="same")(x)
    x = layers.Conv2D(64, 3, padding="same", activation=tf.nn.leaky_relu)(x)

    outputs = layers.Conv2D(num_classes, 1, activation="sigmoid", dtype="float32")(x)
    return models.Model(inputs, outputs, name="TransUNet")

# ===============================================
# 6. Metrics Function
# ===============================================
def calculate_metrics(y_true, y_pred, threshold=0.5, smooth=1e-6):
    y_pred_bin = (y_pred > threshold).astype(np.float32)
    pixel_acc = np.mean(y_true == y_pred_bin)
    intersection = np.sum(y_true * y_pred_bin)
    dice = (2 * intersection + smooth) / (np.sum(y_true) + np.sum(y_pred_bin) + smooth)
    union = np.sum(y_true) + np.sum(y_pred_bin) - intersection
    iou = (intersection + smooth) / (union + smooth)
    return pixel_acc, dice, iou

# ===============================================
# 7. K-Fold Cross Validation
# ===============================================
from sklearn.utils import shuffle
images, masks = shuffle(images, masks, random_state=42)

num_folds = 5
kf = KFold(n_splits=num_folds, shuffle=True, random_state=42)

fold_metrics = []

for fold, (train_idx, val_idx) in enumerate(kf.split(images)):
    print(f"\n📂 Fold {fold+1}/{num_folds}")

    X_train, X_val = images[train_idx], images[val_idx]
    y_train, y_val = masks[train_idx], masks[val_idx]

    # Build fresh model
    model = build_transunet(input_shape=(512, 512, 3))
    model.compile(
        optimizer=tf.keras.optimizers.Adam(1e-4),
        loss=bce_dice_loss,
        metrics=["accuracy"]
    )

    # Callbacks
    callbacks = [
        tf.keras.callbacks.ModelCheckpoint(
            f"transunet_fold{fold+1}.h5",
            save_best_only=True,
            monitor="val_loss"
        )
    ]

    # Train
    history = model.fit(
        X_train, y_train,
        validation_data=(X_val, y_val),
        epochs=50,
        batch_size=4,
        callbacks=callbacks,
        verbose=1
    )

    # Evaluate
    preds_val = model.predict(X_val)
    pixel_acc, dice_score, iou_score = calculate_metrics(y_val, preds_val)

    print(f"✅ Fold {fold+1} Results:")
    print(f"   Pixel Accuracy: {pixel_acc:.4f}")
    print(f"   Dice Score:     {dice_score:.4f}")
    print(f"   IoU Score:      {iou_score:.4f}")

    fold_metrics.append((pixel_acc, dice_score, iou_score))

# ===============================================
# 8. Average Metrics Across Folds
# ===============================================
fold_metrics = np.array(fold_metrics)
mean_metrics = np.mean(fold_metrics, axis=0)
std_metrics = np.std(fold_metrics, axis=0)

print("\n📊 Final Cross-Validation Results:")
print(f"Pixel Accuracy: {mean_metrics[0]:.4f} ± {std_metrics[0]:.4f}")
print(f"Dice Score:     {mean_metrics[1]:.4f} ± {std_metrics[1]:.4f}")
print(f"IoU Score:      {mean_metrics[2]:.4f} ± {std_metrics[2]:.4f}")
```

```python
# ===============================================
# 0. Imports
# ===============================================
import os
import zipfile
from glob import glob
import cv2
import numpy as np
import matplotlib.pyplot as plt
from tqdm import tqdm
import tensorflow as tf
from tensorflow.keras import layers, models
from sklearn.model_selection import KFold
import tensorflow.keras.backend as K

# ✅ Enable mixed precision
from tensorflow.keras import mixed_precision
mixed_precision.set_global_policy("mixed_float16")

# ===============================================
# 1. Unzip Dataset + Auto Path Fix
# ===============================================
zip_path = "/content/stage1_train.zip"
extract_root = "/content/stage1_train"
os.makedirs(extract_root, exist_ok=True)

if not any(os.path.isdir(os.path.join(extract_root, d)) for d in os.listdir(extract_root)):
    print("📦 Extracting dataset...")
    with zipfile.ZipFile(zip_path, 'r') as zip_ref:
        zip_ref.extractall(extract_root)
    print("✅ Extraction completed.")
else:
    print("✅ Dataset already extracted.")

candidates = [d for d in glob(os.path.join(extract_root, "*")) if os.path.isdir(d)]
if len(candidates) == 0:
    raise FileNotFoundError("❌ Could not find extracted stage1_train folder.")

if len(candidates) == 1 and "stage1_train" in os.path.basename(candidates[0]):
    extract_path = candidates[0]
else:
    extract_path = extract_root

print("✅ Using dataset folder:", extract_path)
print("🔍 First 5 contents:", os.listdir(extract_path)[:5])

# ===============================================
# 2. Load Data Function (Recursive)
# ===============================================
def load_data(data_dir, img_size=(512, 512)):
    images, masks = [], []
    all_ids = [d for d in glob(os.path.join(data_dir, "*")) if os.path.isdir(d)]
    print(f"🔍 Found {len(all_ids)} patient folders")

    for img_id in tqdm(all_ids, desc="Loading patients"):
        img_path_list = glob(os.path.join(img_id, "images", "*.png"))
        mask_paths = glob(os.path.join(img_id, "masks", "*.png"))

        if len(img_path_list) == 0 or len(mask_paths) == 0:
            continue

        image = cv2.imread(img_path_list[0], cv2.IMREAD_COLOR)
        if image is None:
            continue
        image = cv2.resize(image, img_size)
        image = image / 255.0

        mask = np.zeros(img_size, dtype=np.uint8)
        for mp in mask_paths:
            m = cv2.imread(mp, cv2.IMREAD_GRAYSCALE)
            if m is None:
                continue
            m = cv2.resize(m, img_size)
            mask = np.maximum(mask, m)
        mask = (mask > 127).astype(np.float32)
        mask = np.expand_dims(mask, axis=-1)

        images.append(image)
        masks.append(mask)

    return np.array(images, dtype=np.float32), np.array(masks, dtype=np.float32)

# ===============================================
# 3. Load Dataset
# ===============================================
IMG_SIZE = (512, 512)
images, masks = load_data(extract_path, img_size=IMG_SIZE)
print("✅ Images:", images.shape)
print("✅ Masks:", masks.shape)

if len(images) == 0:
    raise RuntimeError("❌ No images loaded!")

idx = np.random.randint(0, len(images))
plt.subplot(1, 2, 1); plt.imshow(images[idx]); plt.title("Image"); plt.axis("off")
plt.subplot(1, 2, 2); plt.imshow(masks[idx].squeeze(), cmap="gray"); plt.title("Mask"); plt.axis("off")
plt.show()

# ===============================================
# 4. Loss Functions
# ===============================================
def dice_loss(y_true, y_pred, smooth=1e-6):
    y_true_f = K.flatten(y_true)
    y_pred_f = K.flatten(y_pred)
    intersection = K.sum(y_true_f * y_pred_f)
    return 1 - (2. * intersection + smooth) / (K.sum(y_true_f) + K.sum(y_pred_f) + smooth)

def bce_dice_loss(y_true, y_pred):
    bce = tf.keras.losses.binary_crossentropy(y_true, y_pred)
    dice = dice_loss(y_true, y_pred)
    return bce + dice

# ===============================================
# 5. TransUNet Model
# ===============================================
class PatchEmbedding(layers.Layer):
    def __init__(self, embed_dim=768):
        super().__init__()
        self.proj = layers.Conv2D(embed_dim, kernel_size=1, strides=1, padding="valid")
        self.flatten = layers.Reshape((-1, embed_dim))

    def call(self, x):
        x = self.proj(x)
        x = self.flatten(x)
        return x

class PositionalEncoding(layers.Layer):
    def __init__(self, num_patches, embed_dim):
        super().__init__()
        self.pos_embed = self.add_weight(
            name="pos_embed",
            shape=(1, num_patches, embed_dim),
            initializer="random_normal"
        )

    def call(self, x):
        return x + self.pos_embed

class TransformerBlock(layers.Layer):
    def __init__(self, embed_dim=768, num_heads=12, mlp_dim=3072, dropout=0.1):
        super().__init__()
        self.norm1 = layers.LayerNormalization(epsilon=1e-6)
        self.attn = layers.MultiHeadAttention(num_heads=num_heads, key_dim=embed_dim, dropout=dropout)
        self.drop1 = layers.Dropout(dropout)
        self.norm2 = layers.LayerNormalization(epsilon=1e-6)
        self.mlp = models.Sequential([
            layers.Dense(mlp_dim, activation=tf.nn.gelu),
            layers.Dropout(dropout),
            layers.Dense(embed_dim),
            layers.Dropout(dropout),
        ])

    def call(self, x):
        attn_output = self.attn(self.norm1(x), self.norm1(x))
        x = x + self.drop1(attn_output)
        mlp_output = self.mlp(self.norm2(x))
        return x + mlp_output

def build_transformer_encoder(x, num_layers=4, embed_dim=768, num_heads=12, mlp_dim=3072):
    for _ in range(num_layers):
        x = TransformerBlock(embed_dim, num_heads, mlp_dim)(x)
    return x

def build_transunet(input_shape=(512, 512, 3), num_classes=1):
    inputs = layers.Input(shape=input_shape)

    base_model = tf.keras.applications.ResNet50(
        include_top=False,
        weights="imagenet",
        input_tensor=inputs
    )

    skip1 = base_model.get_layer("conv1_relu").output          # 256x256
    skip2 = base_model.get_layer("conv2_block3_out").output    # 128x128
    skip3 = base_model.get_layer("conv3_block4_out").output    # 64x64
    cnn_feature = base_model.get_layer("conv4_block6_out").output  # 32x32

    # Tokenize features
    patches = PatchEmbedding(embed_dim=768)(cnn_feature)

    # Add positional encoding
    num_patches = patches.shape[1]
    patches = PositionalEncoding(num_patches, 768)(patches)

    # Transformer encoder
    transformer_out = build_transformer_encoder(patches)

    # Reshape back to feature map
    h, w = cnn_feature.shape[1], cnn_feature.shape[2]
    x = layers.Reshape((h, w, 768))(transformer_out)

    # Decoder
    x = layers.Conv2DTranspose(512, 3, strides=2, padding="same")(x)
    x = layers.Concatenate()([x, skip3])
    x = layers.Conv2D(512, 3, padding="same", activation=tf.nn.leaky_relu)(x)

    x = layers.Conv2DTranspose(256, 3, strides=2, padding="same")(x)
    x = layers.Concatenate()([x, skip2])
    x = layers.Conv2D(256, 3, padding="same", activation=tf.nn.leaky_relu)(x)

    x = layers.Conv2DTranspose(128, 3, strides=2, padding="same")(x)
    x = layers.Concatenate()([x, skip1])
    x = layers.Conv2D(128, 3, padding="same", activation=tf.nn.leaky_relu)(x)

    x = layers.Conv2DTranspose(64, 3, strides=2, padding="same")(x)
    x = layers.Conv2D(64, 3, padding="same", activation=tf.nn.leaky_relu)(x)

    outputs = layers.Conv2D(num_classes, 1, activation="sigmoid", dtype="float32")(x)
    return models.Model(inputs, outputs, name="TransUNet")

# ===============================================
# 6. Metrics Function
# ===============================================
def calculate_metrics(y_true, y_pred, threshold=0.5, smooth=1e-6):
    y_pred_bin = (y_pred > threshold).astype(np.float32)
    pixel_acc = np.mean(y_true == y_pred_bin)
    intersection = np.sum(y_true * y_pred_bin)
    dice = (2 * intersection + smooth) / (np.sum(y_true) + np.sum(y_pred_bin) + smooth)
    union = np.sum(y_true) + np.sum(y_pred_bin) - intersection
    iou = (intersection + smooth) / (union + smooth)
    return pixel_acc, dice, iou

# ===============================================
# 7. K-Fold Cross Validation
# ===============================================
from sklearn.utils import shuffle
images, masks = shuffle(images, masks, random_state=42)

num_folds = 5
kf = KFold(n_splits=num_folds, shuffle=True, random_state=42)

fold_metrics = []

for fold, (train_idx, val_idx) in enumerate(kf.split(images)):
    print(f"\n📂 Fold {fold+1}/{num_folds}")

    X_train, X_val = images[train_idx], images[val_idx]
    y_train, y_val = masks[train_idx], masks[val_idx]

    # Build fresh model
    model = build_transunet(input_shape=(512, 512, 3))
    model.compile(
        optimizer=tf.keras.optimizers.Adam(1e-4),
        loss=bce_dice_loss,
        metrics=["accuracy"]
    )

    # Callbacks
    callbacks = [
        tf.keras.callbacks.ModelCheckpoint(
            f"transunet_16_fold{fold+1}.h5",
            save_best_only=True,
            monitor="val_loss"
        )
    ]

    # Train
    history = model.fit(
        X_train, y_train,
        validation_data=(X_val, y_val),
        epochs=50,
        batch_size=16,
        callbacks=callbacks,
        verbose=1
    )

    # Evaluate
    preds_val = model.predict(X_val)
    pixel_acc, dice_score, iou_score = calculate_metrics(y_val, preds_val)

    print(f"✅ Fold {fold+1} Results:")
    print(f"   Pixel Accuracy: {pixel_acc:.4f}")
    print(f"   Dice Score:     {dice_score:.4f}")
    print(f"   IoU Score:      {iou_score:.4f}")

    fold_metrics.append((pixel_acc, dice_score, iou_score))

# ===============================================
# 8. Average Metrics Across Folds
# ===============================================
fold_metrics = np.array(fold_metrics)
mean_metrics = np.mean(fold_metrics, axis=0)
std_metrics = np.std(fold_metrics, axis=0)

print("\n📊 Final Cross-Validation Results:")
print(f"Pixel Accuracy: {mean_metrics[0]:.4f} ± {std_metrics[0]:.4f}")
print(f"Dice Score:     {mean_metrics[1]:.4f} ± {std_metrics[1]:.4f}")
print(f"IoU Score:      {mean_metrics[2]:.4f} ± {std_metrics[2]:.4f}")
```

```python
# ===============================================
# 0. Imports
# ===============================================
import os
import zipfile
from glob import glob
import cv2
import numpy as np
import matplotlib.pyplot as plt
from tqdm import tqdm
import tensorflow as tf
from tensorflow.keras import layers, models
from sklearn.model_selection import KFold
import tensorflow.keras.backend as K

# ✅ Enable mixed precision
from tensorflow.keras import mixed_precision
mixed_precision.set_global_policy("mixed_float16")

# ===============================================
# 1. Unzip Dataset + Auto Path Fix
# ===============================================
zip_path = "/content/stage1_train.zip"
extract_root = "/content/stage1_train"
os.makedirs(extract_root, exist_ok=True)

if not any(os.path.isdir(os.path.join(extract_root, d)) for d in os.listdir(extract_root)):
    print("📦 Extracting dataset...")
    with zipfile.ZipFile(zip_path, 'r') as zip_ref:
        zip_ref.extractall(extract_root)
    print("✅ Extraction completed.")
else:
    print("✅ Dataset already extracted.")

candidates = [d for d in glob(os.path.join(extract_root, "*")) if os.path.isdir(d)]
if len(candidates) == 0:
    raise FileNotFoundError("❌ Could not find extracted stage1_train folder.")

if len(candidates) == 1 and "stage1_train" in os.path.basename(candidates[0]):
    extract_path = candidates[0]
else:
    extract_path = extract_root

print("✅ Using dataset folder:", extract_path)
print("🔍 First 5 contents:", os.listdir(extract_path)[:5])

# ===============================================
# 2. Load Data Function (Recursive)
# ===============================================
def load_data(data_dir, img_size=(512, 512)):
    images, masks = [], []
    all_ids = [d for d in glob(os.path.join(data_dir, "*")) if os.path.isdir(d)]
    print(f"🔍 Found {len(all_ids)} patient folders")

    for img_id in tqdm(all_ids, desc="Loading patients"):
        img_path_list = glob(os.path.join(img_id, "images", "*.png"))
        mask_paths = glob(os.path.join(img_id, "masks", "*.png"))

        if len(img_path_list) == 0 or len(mask_paths) == 0:
            continue

        image = cv2.imread(img_path_list[0], cv2.IMREAD_COLOR)
        if image is None:
            continue
        image = cv2.resize(image, img_size)
        image = image / 255.0

        mask = np.zeros(img_size, dtype=np.uint8)
        for mp in mask_paths:
            m = cv2.imread(mp, cv2.IMREAD_GRAYSCALE)
            if m is None:
                continue
            m = cv2.resize(m, img_size)
            mask = np.maximum(mask, m)
        mask = (mask > 127).astype(np.float32)
        mask = np.expand_dims(mask, axis=-1)

        images.append(image)
        masks.append(mask)

    return np.array(images, dtype=np.float32), np.array(masks, dtype=np.float32)

# ===============================================
# 3. Load Dataset
# ===============================================
IMG_SIZE = (512, 512)
images, masks = load_data(extract_path, img_size=IMG_SIZE)
print("✅ Images:", images.shape)
print("✅ Masks:", masks.shape)

if len(images) == 0:
    raise RuntimeError("❌ No images loaded!")

idx = np.random.randint(0, len(images))
plt.subplot(1, 2, 1); plt.imshow(images[idx]); plt.title("Image"); plt.axis("off")
plt.subplot(1, 2, 2); plt.imshow(masks[idx].squeeze(), cmap="gray"); plt.title("Mask"); plt.axis("off")
plt.show()

# ===============================================
# 4. Loss Functions
# ===============================================
def dice_loss(y_true, y_pred, smooth=1e-6):
    y_true_f = K.flatten(y_true)
    y_pred_f = K.flatten(y_pred)
    intersection = K.sum(y_true_f * y_pred_f)
    return 1 - (2. * intersection + smooth) / (K.sum(y_true_f) + K.sum(y_pred_f) + smooth)

def bce_dice_loss(y_true, y_pred):
    bce = tf.keras.losses.binary_crossentropy(y_true, y_pred)
    dice = dice_loss(y_true, y_pred)
    return bce + dice

# ===============================================
# 5. TransUNet Model
# ===============================================
class PatchEmbedding(layers.Layer):
    def __init__(self, embed_dim=768):
        super().__init__()
        self.proj = layers.Conv2D(embed_dim, kernel_size=1, strides=1, padding="valid")
        self.flatten = layers.Reshape((-1, embed_dim))

    def call(self, x):
        x = self.proj(x)
        x = self.flatten(x)
        return x

class PositionalEncoding(layers.Layer):
    def __init__(self, num_patches, embed_dim):
        super().__init__()
        self.pos_embed = self.add_weight(
            name="pos_embed",
            shape=(1, num_patches, embed_dim),
            initializer="random_normal"
        )

    def call(self, x):
        return x + self.pos_embed

class TransformerBlock(layers.Layer):
    def __init__(self, embed_dim=768, num_heads=12, mlp_dim=3072, dropout=0.1):
        super().__init__()
        self.norm1 = layers.LayerNormalization(epsilon=1e-6)
        self.attn = layers.MultiHeadAttention(num_heads=num_heads, key_dim=embed_dim, dropout=dropout)
        self.drop1 = layers.Dropout(dropout)
        self.norm2 = layers.LayerNormalization(epsilon=1e-6)
        self.mlp = models.Sequential([
            layers.Dense(mlp_dim, activation=tf.nn.gelu),
            layers.Dropout(dropout),
            layers.Dense(embed_dim),
            layers.Dropout(dropout),
        ])

    def call(self, x):
        attn_output = self.attn(self.norm1(x), self.norm1(x))
        x = x + self.drop1(attn_output)
        mlp_output = self.mlp(self.norm2(x))
        return x + mlp_output

def build_transformer_encoder(x, num_layers=4, embed_dim=768, num_heads=8, mlp_dim=3072):
    for _ in range(num_layers):
        x = TransformerBlock(embed_dim, num_heads, mlp_dim)(x)
    return x

def build_transunet(input_shape=(512, 512, 3), num_classes=1):
    inputs = layers.Input(shape=input_shape)

    base_model = tf.keras.applications.ResNet50(
        include_top=False,
        weights="imagenet",
        input_tensor=inputs
    )

    skip1 = base_model.get_layer("conv1_relu").output          # 256x256
    skip2 = base_model.get_layer("conv2_block3_out").output    # 128x128
    skip3 = base_model.get_layer("conv3_block4_out").output    # 64x64
    cnn_feature = base_model.get_layer("conv4_block6_out").output  # 32x32

    # Tokenize features
    patches = PatchEmbedding(embed_dim=768)(cnn_feature)

    # Add positional encoding
    num_patches = patches.shape[1]
    patches = PositionalEncoding(num_patches, 768)(patches)

    # Transformer encoder
    transformer_out = build_transformer_encoder(patches)

    # Reshape back to feature map
    h, w = cnn_feature.shape[1], cnn_feature.shape[2]
    x = layers.Reshape((h, w, 768))(transformer_out)

    # Decoder
    x = layers.Conv2DTranspose(512, 3, strides=2, padding="same")(x)
    x = layers.Concatenate()([x, skip3])
    x = layers.Conv2D(512, 3, padding="same", activation=tf.nn.leaky_relu)(x)

    x = layers.Conv2DTranspose(256, 3, strides=2, padding="same")(x)
    x = layers.Concatenate()([x, skip2])
    x = layers.Conv2D(256, 3, padding="same", activation=tf.nn.leaky_relu)(x)

    x = layers.Conv2DTranspose(128, 3, strides=2, padding="same")(x)
    x = layers.Concatenate()([x, skip1])
    x = layers.Conv2D(128, 3, padding="same", activation=tf.nn.leaky_relu)(x)

    x = layers.Conv2DTranspose(64, 3, strides=2, padding="same")(x)
    x = layers.Conv2D(64, 3, padding="same", activation=tf.nn.leaky_relu)(x)

    outputs = layers.Conv2D(num_classes, 1, activation="sigmoid", dtype="float32")(x)
    return models.Model(inputs, outputs, name="TransUNet")

# ===============================================
# 6. Metrics Function
# ===============================================
def calculate_metrics(y_true, y_pred, threshold=0.5, smooth=1e-6):
    y_pred_bin = (y_pred > threshold).astype(np.float32)
    pixel_acc = np.mean(y_true == y_pred_bin)
    intersection = np.sum(y_true * y_pred_bin)
    dice = (2 * intersection + smooth) / (np.sum(y_true) + np.sum(y_pred_bin) + smooth)
    union = np.sum(y_true) + np.sum(y_pred_bin) - intersection
    iou = (intersection + smooth) / (union + smooth)
    return pixel_acc, dice, iou

# ===============================================
# 7. K-Fold Cross Validation
# ===============================================
from sklearn.utils import shuffle
images, masks = shuffle(images, masks, random_state=42)

num_folds = 5
kf = KFold(n_splits=num_folds, shuffle=True, random_state=42)

fold_metrics = []

for fold, (train_idx, val_idx) in enumerate(kf.split(images)):
    print(f"\n📂 Fold {fold+1}/{num_folds}")

    X_train, X_val = images[train_idx], images[val_idx]
    y_train, y_val = masks[train_idx], masks[val_idx]

    # Build fresh model
    model = build_transunet(input_shape=(512, 512, 3))
    model.compile(
        optimizer=tf.keras.optimizers.Adam(1e-4),
        loss=bce_dice_loss,
        metrics=["accuracy"]
    )

    # Callbacks
    callbacks = [
        tf.keras.callbacks.ModelCheckpoint(
            f"transunet_16_fold{fold+1}.h5",
            save_best_only=True,
            monitor="val_loss"
        )
    ]

    # Train
    history = model.fit(
        X_train, y_train,
        validation_data=(X_val, y_val),
        epochs=50,
        batch_size=16,
        callbacks=callbacks,
        verbose=1
    )

    # Evaluate
    preds_val = model.predict(X_val)
    pixel_acc, dice_score, iou_score = calculate_metrics(y_val, preds_val)

    print(f"✅ Fold {fold+1} Results:")
    print(f"   Pixel Accuracy: {pixel_acc:.4f}")
    print(f"   Dice Score:     {dice_score:.4f}")
    print(f"   IoU Score:      {iou_score:.4f}")

    fold_metrics.append((pixel_acc, dice_score, iou_score))

# ===============================================
# 8. Average Metrics Across Folds
# ===============================================
fold_metrics = np.array(fold_metrics)
mean_metrics = np.mean(fold_metrics, axis=0)
std_metrics = np.std(fold_metrics, axis=0)

print("\n📊 Final Cross-Validation Results:")
print(f"Pixel Accuracy: {mean_metrics[0]:.4f} ± {std_metrics[0]:.4f}")
print(f"Dice Score:     {mean_metrics[1]:.4f} ± {std_metrics[1]:.4f}")
print(f"IoU Score:      {mean_metrics[2]:.4f} ± {std_metrics[2]:.4f}")
```

```python
# ===============================================
# 0. Imports
# ===============================================
import os
import zipfile
from glob import glob
import cv2
import numpy as np
import matplotlib.pyplot as plt
from tqdm import tqdm
import tensorflow as tf
from tensorflow.keras import layers, models
from sklearn.model_selection import KFold
import tensorflow.keras.backend as K

# ✅ Enable mixed precision
from tensorflow.keras import mixed_precision
mixed_precision.set_global_policy("mixed_float16")

# ===============================================
# 1. Unzip Dataset + Auto Path Fix
# ===============================================
zip_path = "/content/stage1_train.zip"
extract_root = "/content/stage1_train"
os.makedirs(extract_root, exist_ok=True)

if not any(os.path.isdir(os.path.join(extract_root, d)) for d in os.listdir(extract_root)):
    print("📦 Extracting dataset...")
    with zipfile.ZipFile(zip_path, 'r') as zip_ref:
        zip_ref.extractall(extract_root)
    print("✅ Extraction completed.")
else:
    print("✅ Dataset already extracted.")

candidates = [d for d in glob(os.path.join(extract_root, "*")) if os.path.isdir(d)]
if len(candidates) == 0:
    raise FileNotFoundError("❌ Could not find extracted stage1_train folder.")

if len(candidates) == 1 and "stage1_train" in os.path.basename(candidates[0]):
    extract_path = candidates[0]
else:
    extract_path = extract_root

print("✅ Using dataset folder:", extract_path)
print("🔍 First 5 contents:", os.listdir(extract_path)[:5])

# ===============================================
# 2. Load Data Function (Recursive)
# ===============================================
def load_data(data_dir, img_size=(512, 512)):
    images, masks = [], []
    all_ids = [d for d in glob(os.path.join(data_dir, "*")) if os.path.isdir(d)]
    print(f"🔍 Found {len(all_ids)} patient folders")

    for img_id in tqdm(all_ids, desc="Loading patients"):
        img_path_list = glob(os.path.join(img_id, "images", "*.png"))
        mask_paths = glob(os.path.join(img_id, "masks", "*.png"))

        if len(img_path_list) == 0 or len(mask_paths) == 0:
            continue

        image = cv2.imread(img_path_list[0], cv2.IMREAD_COLOR)
        if image is None:
            continue
        image = cv2.resize(image, img_size)
        image = image / 255.0

        mask = np.zeros(img_size, dtype=np.uint8)
        for mp in mask_paths:
            m = cv2.imread(mp, cv2.IMREAD_GRAYSCALE)
            if m is None:
                continue
            m = cv2.resize(m, img_size)
            mask = np.maximum(mask, m)
        mask = (mask > 127).astype(np.float32)
        mask = np.expand_dims(mask, axis=-1)

        images.append(image)
        masks.append(mask)

    return np.array(images, dtype=np.float32), np.array(masks, dtype=np.float32)

# ===============================================
# 3. Load Dataset
# ===============================================
IMG_SIZE = (512, 512)
images, masks = load_data(extract_path, img_size=IMG_SIZE)
print("✅ Images:", images.shape)
print("✅ Masks:", masks.shape)

if len(images) == 0:
    raise RuntimeError("❌ No images loaded!")

idx = np.random.randint(0, len(images))
plt.subplot(1, 2, 1); plt.imshow(images[idx]); plt.title("Image"); plt.axis("off")
plt.subplot(1, 2, 2); plt.imshow(masks[idx].squeeze(), cmap="gray"); plt.title("Mask"); plt.axis("off")
plt.show()

# ===============================================
# 4. Loss Functions
# ===============================================
def dice_loss(y_true, y_pred, smooth=1e-6):
    y_true_f = K.flatten(y_true)
    y_pred_f = K.flatten(y_pred)
    intersection = K.sum(y_true_f * y_pred_f)
    return 1 - (2. * intersection + smooth) / (K.sum(y_true_f) + K.sum(y_pred_f) + smooth)

def bce_dice_loss(y_true, y_pred):
    bce = tf.keras.losses.binary_crossentropy(y_true, y_pred)
    dice = dice_loss(y_true, y_pred)
    return bce + dice

# ===============================================
# 5. TransUNet Model
# ===============================================
class PatchEmbedding(layers.Layer):
    def __init__(self, embed_dim=768):
        super().__init__()
        self.proj = layers.Conv2D(embed_dim, kernel_size=1, strides=1, padding="valid")
        self.flatten = layers.Reshape((-1, embed_dim))

    def call(self, x):
        x = self.proj(x)
        x = self.flatten(x)
        return x

class PositionalEncoding(layers.Layer):
    def __init__(self, num_patches, embed_dim):
        super().__init__()
        self.pos_embed = self.add_weight(
            name="pos_embed",
            shape=(1, num_patches, embed_dim),
            initializer="random_normal"
        )

    def call(self, x):
        return x + self.pos_embed

class TransformerBlock(layers.Layer):
    def __init__(self, embed_dim=768, num_heads=12, mlp_dim=3072, dropout=0.1):
        super().__init__()
        self.norm1 = layers.LayerNormalization(epsilon=1e-6)
        self.attn = layers.MultiHeadAttention(num_heads=num_heads, key_dim=embed_dim, dropout=dropout)
        self.drop1 = layers.Dropout(dropout)
        self.norm2 = layers.LayerNormalization(epsilon=1e-6)
        self.mlp = models.Sequential([
            layers.Dense(mlp_dim, activation=tf.nn.gelu),
            layers.Dropout(dropout),
            layers.Dense(embed_dim),
            layers.Dropout(dropout),
        ])

    def call(self, x):
        attn_output = self.attn(self.norm1(x), self.norm1(x))
        x = x + self.drop1(attn_output)
        mlp_output = self.mlp(self.norm2(x))
        return x + mlp_output

def build_transformer_encoder(x, num_layers=4, embed_dim=768, num_heads=12, mlp_dim=3072):
    for _ in range(num_layers):
        x = TransformerBlock(embed_dim, num_heads, mlp_dim)(x)
    return x

def build_transunet(input_shape=(512, 512, 3), num_classes=1):
    inputs = layers.Input(shape=input_shape)

    base_model = tf.keras.applications.ResNet50(
        include_top=False,
        weights="imagenet",
        input_tensor=inputs
    )

    skip1 = base_model.get_layer("conv1_relu").output          # 256x256
    skip2 = base_model.get_layer("conv2_block3_out").output    # 128x128
    skip3 = base_model.get_layer("conv3_block4_out").output    # 64x64
    cnn_feature = base_model.get_layer("conv4_block6_out").output  # 32x32

    # Tokenize features
    patches = PatchEmbedding(embed_dim=768)(cnn_feature)

    # Add positional encoding
    num_patches = patches.shape[1]
    patches = PositionalEncoding(num_patches, 768)(patches)

    # Transformer encoder
    transformer_out = build_transformer_encoder(patches)

    # Reshape back to feature map
    h, w = cnn_feature.shape[1], cnn_feature.shape[2]
    x = layers.Reshape((h, w, 768))(transformer_out)

    # Decoder
    x = layers.Conv2DTranspose(512, 3, strides=2, padding="same")(x)
    x = layers.Concatenate()([x, skip3])
    x = layers.Conv2D(512, 3, padding="same", activation=tf.nn.leaky_relu)(x)

    x = layers.Conv2DTranspose(256, 3, strides=2, padding="same")(x)
    x = layers.Concatenate()([x, skip2])
    x = layers.Conv2D(256, 3, padding="same", activation=tf.nn.leaky_relu)(x)

    x = layers.Conv2DTranspose(128, 3, strides=2, padding="same")(x)
    x = layers.Concatenate()([x, skip1])
    x = layers.Conv2D(128, 3, padding="same", activation=tf.nn.leaky_relu)(x)

    x = layers.Conv2DTranspose(64, 3, strides=2, padding="same")(x)
    x = layers.Conv2D(64, 3, padding="same", activation=tf.nn.leaky_relu)(x)

    outputs = layers.Conv2D(num_classes, 1, activation="sigmoid", dtype="float32")(x)
    return models.Model(inputs, outputs, name="TransUNet")

# ===============================================
# 6. Metrics Function
# ===============================================
def calculate_metrics(y_true, y_pred, threshold=0.5, smooth=1e-6):
    y_pred_bin = (y_pred > threshold).astype(np.float32)
    pixel_acc = np.mean(y_true == y_pred_bin)
    intersection = np.sum(y_true * y_pred_bin)
    dice = (2 * intersection + smooth) / (np.sum(y_true) + np.sum(y_pred_bin) + smooth)
    union = np.sum(y_true) + np.sum(y_pred_bin) - intersection
    iou = (intersection + smooth) / (union + smooth)
    return pixel_acc, dice, iou

# ===============================================
# 7. K-Fold Cross Validation
# ===============================================
from sklearn.utils import shuffle
images, masks = shuffle(images, masks, random_state=42)

num_folds = 5
kf = KFold(n_splits=num_folds, shuffle=True, random_state=42)

fold_metrics = []

for fold, (train_idx, val_idx) in enumerate(kf.split(images)):
    print(f"\n📂 Fold {fold+1}/{num_folds}")

    X_train, X_val = images[train_idx], images[val_idx]
    y_train, y_val = masks[train_idx], masks[val_idx]

    # Build fresh model
    model = build_transunet(input_shape=(512, 512, 3))
    model.compile(
        optimizer=tf.keras.optimizers.Adam(1e-4),
        loss=bce_dice_loss,
        metrics=["accuracy"]
    )

    # Callbacks
    callbacks = [
        tf.keras.callbacks.ModelCheckpoint(
            f"transunet_16_fold{fold+1}.weights.h5",
            save_best_only=True,
            monitor="val_loss"
        )
    ]

    # Train
    history = model.fit(
        X_train, y_train,
        validation_data=(X_val, y_val),
        epochs=50,
        batch_size=16,
        callbacks=callbacks,
        verbose=1
    )

    # Evaluate
    preds_val = model.predict(X_val)
    pixel_acc, dice_score, iou_score = calculate_metrics(y_val, preds_val)

    print(f"✅ Fold {fold+1} Results:")
    print(f"   Pixel Accuracy: {pixel_acc:.4f}")
    print(f"   Dice Score:     {dice_score:.4f}")
    print(f"   IoU Score:      {iou_score:.4f}")

    fold_metrics.append((pixel_acc, dice_score, iou_score))

# ===============================================
# 8. Average Metrics Across Folds
# ===============================================
fold_metrics = np.array(fold_metrics)
mean_metrics = np.mean(fold_metrics, axis=0)
std_metrics = np.std(fold_metrics, axis=0)

print("\n📊 Final Cross-Validation Results:")
print(f"Pixel Accuracy: {mean_metrics[0]:.4f} ± {std_metrics[0]:.4f}")
print(f"Dice Score:     {mean_metrics[1]:.4f} ± {std_metrics[1]:.4f}")
print(f"IoU Score:      {mean_metrics[2]:.4f} ± {std_metrics[2]:.4f}")
```

```python
# ===============================================
# 0. Imports
# ===============================================
import os
import zipfile
from glob import glob
import cv2
import numpy as np
import matplotlib.pyplot as plt
from tqdm import tqdm
import tensorflow as tf
from tensorflow.keras import layers, models
from sklearn.model_selection import KFold
import tensorflow.keras.backend as K

# ✅ Enable mixed precision
from tensorflow.keras import mixed_precision
mixed_precision.set_global_policy("mixed_float16")

# ===============================================
# 1. Unzip Dataset + Auto Path Fix
# ===============================================
zip_path = "/content/stage1_train.zip"
extract_root = "/content/stage1_train"
os.makedirs(extract_root, exist_ok=True)

if not any(os.path.isdir(os.path.join(extract_root, d)) for d in os.listdir(extract_root)):
    print("📦 Extracting dataset...")
    with zipfile.ZipFile(zip_path, 'r') as zip_ref:
        zip_ref.extractall(extract_root)
    print("✅ Extraction completed.")
else:
    print("✅ Dataset already extracted.")

candidates = [d for d in glob(os.path.join(extract_root, "*")) if os.path.isdir(d)]
extract_path = candidates[0] if len(candidates) == 1 else extract_root
print("✅ Using dataset folder:", extract_path)
print("🔍 First 5 contents:", os.listdir(extract_path)[:5])

# ===============================================
# 2. Load Data Function (Recursive)
# ===============================================
def load_data(data_dir, img_size=(512, 512)):
    images, masks = [], []
    all_ids = [d for d in glob(os.path.join(data_dir, "*")) if os.path.isdir(d)]
    print(f"🔍 Found {len(all_ids)} patient folders")

    for img_id in tqdm(all_ids, desc="Loading patients"):
        img_path_list = glob(os.path.join(img_id, "images", "*.png"))
        mask_paths = glob(os.path.join(img_id, "masks", "*.png"))

        if len(img_path_list) == 0 or len(mask_paths) == 0:
            continue

        image = cv2.imread(img_path_list[0], cv2.IMREAD_COLOR)
        if image is None:
            continue
        image = cv2.resize(image, img_size)
        image = image / 255.0

        mask = np.zeros(img_size, dtype=np.uint8)
        for mp in mask_paths:
            m = cv2.imread(mp, cv2.IMREAD_GRAYSCALE)
            if m is None:
                continue
            m = cv2.resize(m, img_size)
            mask = np.maximum(mask, m)
        mask = (mask > 127).astype(np.float32)
        mask = np.expand_dims(mask, axis=-1)

        images.append(image)
        masks.append(mask)

    return np.array(images, dtype=np.float32), np.array(masks, dtype=np.float32)

# ===============================================
# 3. Load Dataset
# ===============================================
IMG_SIZE = (512, 512)
images, masks = load_data(extract_path, img_size=IMG_SIZE)
print("✅ Images:", images.shape)
print("✅ Masks:", masks.shape)

idx = np.random.randint(0, len(images))
plt.subplot(1, 2, 1); plt.imshow(images[idx]); plt.title("Image"); plt.axis("off")
plt.subplot(1, 2, 2); plt.imshow(masks[idx].squeeze(), cmap="gray"); plt.title("Mask"); plt.axis("off")
plt.show()

# ===============================================
# 4. Loss Functions
# ===============================================
def dice_loss(y_true, y_pred, smooth=1e-6):
    y_true_f = K.flatten(y_true)
    y_pred_f = K.flatten(y_pred)
    intersection = K.sum(y_true_f * y_pred_f)
    return 1 - (2. * intersection + smooth) / (K.sum(y_true_f) + K.sum(y_pred_f) + smooth)

def bce_dice_loss(y_true, y_pred):
    bce = tf.keras.losses.binary_crossentropy(y_true, y_pred)
    dice = dice_loss(y_true, y_pred)
    return bce + dice

# ===============================================
# 5. TransUNet Model
# ===============================================
class PatchEmbedding(layers.Layer):
    def __init__(self, embed_dim=192):
        super().__init__()
        self.proj = layers.Conv2D(embed_dim, kernel_size=1, strides=1, padding="valid")
        self.flatten = layers.Reshape((-1, embed_dim))

    def call(self, x):
        x = self.proj(x)
        x = self.flatten(x)
        return x

class PositionalEncoding(layers.Layer):
    def __init__(self, num_patches, embed_dim):
        super().__init__()
        self.pos_embed = self.add_weight(
            name="pos_embed",
            shape=(1, num_patches, embed_dim),
            initializer="random_normal"
        )

    def call(self, x):
        return x + self.pos_embed

class TransformerBlock(layers.Layer):
    def __init__(self, embed_dim=192, num_heads=12, mlp_dim=768, dropout=0.1):
        super().__init__()
        self.norm1 = layers.LayerNormalization(epsilon=1e-6)
        self.attn = layers.MultiHeadAttention(num_heads=num_heads, key_dim=embed_dim, dropout=dropout)
        self.drop1 = layers.Dropout(dropout)
        self.norm2 = layers.LayerNormalization(epsilon=1e-6)
        self.mlp = models.Sequential([
            layers.Dense(mlp_dim, activation=tf.nn.gelu),
            layers.Dropout(dropout),
            layers.Dense(embed_dim),
            layers.Dropout(dropout),
        ])

    def call(self, x):
        attn_output = self.attn(self.norm1(x), self.norm1(x))
        x = x + self.drop1(attn_output)
        mlp_output = self.mlp(self.norm2(x))
        return x + mlp_output

def build_transformer_encoder(x, num_layers=4, embed_dim=192, num_heads=12, mlp_dim=768):
    for _ in range(num_layers):
        x = TransformerBlock(embed_dim, num_heads, mlp_dim)(x)
    return x

def build_transunet(input_shape=(512, 512, 3), num_classes=1):
    inputs = layers.Input(shape=input_shape)

    base_model = tf.keras.applications.ResNet50(
        include_top=False,
        weights="imagenet",
        input_tensor=inputs
    )

    skip1 = base_model.get_layer("conv1_relu").output
    skip2 = base_model.get_layer("conv2_block3_out").output
    skip3 = base_model.get_layer("conv3_block4_out").output
    cnn_feature = base_model.get_layer("conv4_block6_out").output

    patches = PatchEmbedding(embed_dim=192)(cnn_feature)
    num_patches = patches.shape[1]
    patches = PositionalEncoding(num_patches, 192)(patches)
    transformer_out = build_transformer_encoder(patches)

    h, w = cnn_feature.shape[1], cnn_feature.shape[2]
    x = layers.Reshape((h, w, 192))(transformer_out)

    # Decoder
    x = layers.Conv2DTranspose(512, 3, strides=2, padding="same")(x)
    x = layers.Concatenate()([x, skip3])
    x = layers.Conv2D(512, 3, padding="same", activation=tf.nn.leaky_relu)(x)

    x = layers.Conv2DTranspose(256, 3, strides=2, padding="same")(x)
    x = layers.Concatenate()([x, skip2])
    x = layers.Conv2D(256, 3, padding="same", activation=tf.nn.leaky_relu)(x)

    x = layers.Conv2DTranspose(128, 3, strides=2, padding="same")(x)
    x = layers.Concatenate()([x, skip1])
    x = layers.Conv2D(128, 3, padding="same", activation=tf.nn.leaky_relu)(x)

    x = layers.Conv2DTranspose(64, 3, strides=2, padding="same")(x)
    x = layers.Conv2D(64, 3, padding="same", activation=tf.nn.leaky_relu)(x)

    outputs = layers.Conv2D(num_classes, 1, activation="sigmoid", dtype="float32")(x)
    return models.Model(inputs, outputs, name="TransUNet")

# ===============================================
# 6. Metrics Function
# ===============================================
def calculate_metrics(y_true, y_pred, threshold=0.5, smooth=1e-6):
    y_pred_bin = (y_pred > threshold).astype(np.float32)
    pixel_acc = np.mean(y_true == y_pred_bin)
    intersection = np.sum(y_true * y_pred_bin)
    dice = (2 * intersection + smooth) / (np.sum(y_true) + np.sum(y_pred_bin) + smooth)
    union = np.sum(y_true) + np.sum(y_pred_bin) - intersection
    iou = (intersection + smooth) / (union + smooth)
    return pixel_acc, dice, iou

# ===============================================
# 7. K-Fold Cross Validation
# ===============================================
from sklearn.utils import shuffle
images, masks = shuffle(images, masks, random_state=42)

num_folds = 5
kf = KFold(n_splits=num_folds, shuffle=True, random_state=42)

fold_metrics = []

for fold, (train_idx, val_idx) in enumerate(kf.split(images)):
    print(f"\n📂 Fold {fold+1}/{num_folds}")

    X_train, X_val = images[train_idx], images[val_idx]
    y_train, y_val = masks[train_idx], masks[val_idx]

    model = build_transunet(input_shape=(512, 512, 3))
    model.compile(
        optimizer=tf.keras.optimizers.Adam(1e-4),
        loss=bce_dice_loss,
        metrics=["accuracy"]
    )

    callbacks = [
        tf.keras.callbacks.ModelCheckpoint(
            f"transunet_768_fold{fold+1}.weights.h5",
            save_best_only=True,
            monitor="val_loss"
        )
    ]

    history = model.fit(
        X_train, y_train,
        validation_data=(X_val, y_val),
        epochs=50,
        batch_size=16,
        callbacks=callbacks,
        verbose=1
    )

    # Evaluate
    preds_val = model.predict(X_val)
    pixel_acc, dice_score, iou_score = calculate_metrics(y_val, preds_val)

    print(f"✅ Fold {fold+1} Results:")
    print(f"   Pixel Accuracy: {pixel_acc:.4f}")
    print(f"   Dice Score:     {dice_score:.4f}")
    print(f"   IoU Score:      {iou_score:.4f}")

    fold_metrics.append((pixel_acc, dice_score, iou_score))

# ===============================================
# 8. Average Metrics Across Folds
# ===============================================
fold_metrics = np.array(fold_metrics)
mean_metrics = np.mean(fold_metrics, axis=0)
std_metrics = np.std(fold_metrics, axis=0)

print("\n📊 Final Cross-Validation Results:")
print(f"Pixel Accuracy: {mean_metrics[0]:.4f} ± {std_metrics[0]:.4f}")
print(f"Dice Score:     {mean_metrics[1]:.4f} ± {std_metrics[1]:.4f}")
print(f"IoU Score:      {mean_metrics[2]:.4f} ± {std_metrics[2]:.4f}")
```

```python
# ===============================================
# 0. Imports
# ===============================================
import os
import zipfile
from glob import glob
import cv2
import numpy as np
import matplotlib.pyplot as plt
from tqdm import tqdm
import tensorflow as tf
from tensorflow.keras import layers, models
from sklearn.model_selection import KFold
import tensorflow.keras.backend as K

# ✅ Enable mixed precision
from tensorflow.keras import mixed_precision
mixed_precision.set_global_policy("mixed_float16")

# ===============================================
# 1. Unzip Dataset + Auto Path Fix
# ===============================================
zip_path = "/content/stage1_train.zip"
extract_root = "/content/stage1_train"
os.makedirs(extract_root, exist_ok=True)

if not any(os.path.isdir(os.path.join(extract_root, d)) for d in os.listdir(extract_root)):
    print("📦 Extracting dataset...")
    with zipfile.ZipFile(zip_path, 'r') as zip_ref:
        zip_ref.extractall(extract_root)
    print("✅ Extraction completed.")
else:
    print("✅ Dataset already extracted.")

candidates = [d for d in glob(os.path.join(extract_root, "*")) if os.path.isdir(d)]
extract_path = candidates[0] if len(candidates) == 1 else extract_root
print("✅ Using dataset folder:", extract_path)
print("🔍 First 5 contents:", os.listdir(extract_path)[:5])

# ===============================================
# 2. Load Data Function (Recursive)
# ===============================================
def load_data(data_dir, img_size=(512, 512)):
    images, masks = [], []
    all_ids = [d for d in glob(os.path.join(data_dir, "*")) if os.path.isdir(d)]
    print(f"🔍 Found {len(all_ids)} patient folders")

    for img_id in tqdm(all_ids, desc="Loading patients"):
        img_path_list = glob(os.path.join(img_id, "images", "*.png"))
        mask_paths = glob(os.path.join(img_id, "masks", "*.png"))

        if len(img_path_list) == 0 or len(mask_paths) == 0:
            continue

        image = cv2.imread(img_path_list[0], cv2.IMREAD_COLOR)
        if image is None:
            continue
        image = cv2.resize(image, img_size)
        image = image / 255.0

        mask = np.zeros(img_size, dtype=np.uint8)
        for mp in mask_paths:
            m = cv2.imread(mp, cv2.IMREAD_GRAYSCALE)
            if m is None:
                continue
            m = cv2.resize(m, img_size)
            mask = np.maximum(mask, m)
        mask = (mask > 127).astype(np.float32)
        mask = np.expand_dims(mask, axis=-1)

        images.append(image)
        masks.append(mask)

    return np.array(images, dtype=np.float32), np.array(masks, dtype=np.float32)

# ===============================================
# 3. Load Dataset
# ===============================================
IMG_SIZE = (512, 512)
images, masks = load_data(extract_path, img_size=IMG_SIZE)
print("✅ Images:", images.shape)
print("✅ Masks:", masks.shape)

idx = np.random.randint(0, len(images))
plt.subplot(1, 2, 1); plt.imshow(images[idx]); plt.title("Image"); plt.axis("off")
plt.subplot(1, 2, 2); plt.imshow(masks[idx].squeeze(), cmap="gray"); plt.title("Mask"); plt.axis("off")
plt.show()

# ===============================================
# 4. Loss Functions
# ===============================================
def dice_loss(y_true, y_pred, smooth=1e-6):
    y_true_f = K.flatten(y_true)
    y_pred_f = K.flatten(y_pred)
    intersection = K.sum(y_true_f * y_pred_f)
    return 1 - (2. * intersection + smooth) / (K.sum(y_true_f) + K.sum(y_pred_f) + smooth)

def bce_dice_loss(y_true, y_pred):
    bce = tf.keras.losses.binary_crossentropy(y_true, y_pred)
    dice = dice_loss(y_true, y_pred)
    return bce + dice

# ===============================================
# 5. TransUNet Model (embed_dim=48, mlp_dim=192)
# ===============================================
class PatchEmbedding(layers.Layer):
    def __init__(self, embed_dim=48):
        super().__init__()
        self.proj = layers.Conv2D(embed_dim, kernel_size=1, strides=1, padding="valid")
        self.flatten = layers.Reshape((-1, embed_dim))

    def call(self, x):
        x = self.proj(x)
        x = self.flatten(x)
        return x

class PositionalEncoding(layers.Layer):
    def __init__(self, num_patches, embed_dim):
        super().__init__()
        self.pos_embed = self.add_weight(
            name="pos_embed",
            shape=(1, num_patches, embed_dim),
            initializer="random_normal"
        )

    def call(self, x):
        return x + self.pos_embed

class TransformerBlock(layers.Layer):
    def __init__(self, embed_dim=48, num_heads=12, mlp_dim=192, dropout=0.1):
        super().__init__()
        self.norm1 = layers.LayerNormalization(epsilon=1e-6)
        self.attn = layers.MultiHeadAttention(num_heads=num_heads, key_dim=embed_dim, dropout=dropout)
        self.drop1 = layers.Dropout(dropout)
        self.norm2 = layers.LayerNormalization(epsilon=1e-6)
        self.mlp = models.Sequential([
            layers.Dense(mlp_dim, activation=tf.nn.gelu),
            layers.Dropout(dropout),
            layers.Dense(embed_dim),
            layers.Dropout(dropout),
        ])

    def call(self, x):
        attn_output = self.attn(self.norm1(x), self.norm1(x))
        x = x + self.drop1(attn_output)
        mlp_output = self.mlp(self.norm2(x))
        return x + mlp_output

def build_transformer_encoder(x, num_layers=4, embed_dim=48, num_heads=12, mlp_dim=192):
    for _ in range(num_layers):
        x = TransformerBlock(embed_dim, num_heads, mlp_dim)(x)
    return x

def build_transunet(input_shape=(512, 512, 3), num_classes=1):
    inputs = layers.Input(shape=input_shape)

    base_model = tf.keras.applications.ResNet50(
        include_top=False,
        weights="imagenet",
        input_tensor=inputs
    )

    skip1 = base_model.get_layer("conv1_relu").output
    skip2 = base_model.get_layer("conv2_block3_out").output
    skip3 = base_model.get_layer("conv3_block4_out").output
    cnn_feature = base_model.get_layer("conv4_block6_out").output

    patches = PatchEmbedding(embed_dim=48)(cnn_feature)
    num_patches = patches.shape[1]
    patches = PositionalEncoding(num_patches, 48)(patches)
    transformer_out = build_transformer_encoder(patches, embed_dim=48, mlp_dim=192)

    h, w = cnn_feature.shape[1], cnn_feature.shape[2]
    x = layers.Reshape((h, w, 48))(transformer_out)

    # Decoder
    x = layers.Conv2DTranspose(512, 3, strides=2, padding="same")(x)
    x = layers.Concatenate()([x, skip3])
    x = layers.Conv2D(512, 3, padding="same", activation=tf.nn.leaky_relu)(x)

    x = layers.Conv2DTranspose(256, 3, strides=2, padding="same")(x)
    x = layers.Concatenate()([x, skip2])
    x = layers.Conv2D(256, 3, padding="same", activation=tf.nn.leaky_relu)(x)

    x = layers.Conv2DTranspose(128, 3, strides=2, padding="same")(x)
    x = layers.Concatenate()([x, skip1])
    x = layers.Conv2D(128, 3, padding="same", activation=tf.nn.leaky_relu)(x)

    x = layers.Conv2DTranspose(64, 3, strides=2, padding="same")(x)
    x = layers.Conv2D(64, 3, padding="same", activation=tf.nn.leaky_relu)(x)

    outputs = layers.Conv2D(num_classes, 1, activation="sigmoid", dtype="float32")(x)
    return models.Model(inputs, outputs, name="TransUNet_48_192")

# ===============================================
# 6. Metrics Function
# ===============================================
def calculate_metrics(y_true, y_pred, threshold=0.5, smooth=1e-6):
    y_pred_bin = (y_pred > threshold).astype(np.float32)
    pixel_acc = np.mean(y_true == y_pred_bin)
    intersection = np.sum(y_true * y_pred_bin)
    dice = (2 * intersection + smooth) / (np.sum(y_true) + np.sum(y_pred_bin) + smooth)
    union = np.sum(y_true) + np.sum(y_pred_bin) - intersection
    iou = (intersection + smooth) / (union + smooth)
    return pixel_acc, dice, iou

# ===============================================
# 7. K-Fold Cross Validation
# ===============================================
from sklearn.utils import shuffle
images, masks = shuffle(images, masks, random_state=42)

num_folds = 5
kf = KFold(n_splits=num_folds, shuffle=True, random_state=42)

fold_metrics = []

for fold, (train_idx, val_idx) in enumerate(kf.split(images)):
    print(f"\n📂 Fold {fold+1}/{num_folds}")

    X_train, X_val = images[train_idx], images[val_idx]
    y_train, y_val = masks[train_idx], masks[val_idx]

    model = build_transunet(input_shape=(512, 512, 3))
    model.compile(
        optimizer=tf.keras.optimizers.Adam(1e-4),
        loss=bce_dice_loss,
        metrics=["accuracy"]
    )

    callbacks = [
        tf.keras.callbacks.ModelCheckpoint(
            f"transunet_48_192_fold{fold+1}.weights.h5",
            save_best_only=True,
            monitor="val_loss"
        )
    ]

    history = model.fit(
        X_train, y_train,
        validation_data=(X_val, y_val),
        epochs=50,
        batch_size=16,
        callbacks=callbacks,
        verbose=1
    )

    # Evaluate
    preds_val = model.predict(X_val)
    pixel_acc, dice_score, iou_score = calculate_metrics(y_val, preds_val)

    print(f"✅ Fold {fold+1} Results:")
    print(f"   Pixel Accuracy: {pixel_acc:.4f}")
    print(f"   Dice Score:     {dice_score:.4f}")
    print(f"   IoU Score:      {iou_score:.4f}")

    fold_metrics.append((pixel_acc, dice_score, iou_score))

# ===============================================
# 8. Average Metrics Across Folds
# ===============================================
fold_metrics = np.array(fold_metrics)
mean_metrics = np.mean(fold_metrics, axis=0)
std_metrics = np.std(fold_metrics, axis=0)

print("\n📊 Final Cross-Validation Results:")
print(f"Pixel Accuracy: {mean_metrics[0]:.4f} ± {std_metrics[0]:.4f}")
print(f"Dice Score:     {mean_metrics[1]:.4f} ± {std_metrics[1]:.4f}")
print(f"IoU Score:      {mean_metrics[2]:.4f} ± {std_metrics[2]:.4f}")
```
